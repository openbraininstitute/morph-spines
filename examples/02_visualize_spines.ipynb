# Visualization example: neuron section spines

```python
# Imports
from trimesh import util as triutil

from morph_spines import load_morphology_with_spines
```

```python
# Load spine data from morphology-with-spines H5 file
filepath = "./data/morphology_with_spines/864691134884740346_v1.0.h5"
m = load_morphology_with_spines(filepath, load_meshes=True)
```

```python
# Get the spines mesh for a given neuron's section ID and plot them
section_id = 9
mesh = triutil.concatenate(m.spines.spine_meshes_for_section(section_id))
mesh.show()
```

## Soma mesh

The soma mesh is also accessible from the loaded morphology.

```python
# Access and display the soma mesh
soma_mesh = m.soma.soma_mesh
print(f"Soma mesh: {len(soma_mesh.vertices)} vertices, {len(soma_mesh.faces)} faces")
print(f"Soma center: {m.soma.center}")
soma_mesh.show()
```
